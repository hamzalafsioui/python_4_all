<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 1: Statistics & Dynamic Filtering Above the Average</span>
<span style="background-color: #28a745; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Foundational</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Lists • Statistics • Filtering</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Calculate the mean and median of a dataset without external libraries, extract values strictly greater than the average, and apply percentage normalization.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
In basic exercises, you extracted grades above average. Here, we push the logic further: handle empty lists safely, calculate the median for both odd and even lengths, and mathematically normalize elements.

* **Input Example**:  
  <code>notes = [12, 4, 14, 11, 18, 13, 7, 10, 5, 9, 15, 8, 14, 16]</code>
* **Expected Output Example**:  
  <code>Mean: 11.86 | Median: 12.5 | Above average: [12, 14, 14, 18, 13, 15, 16, 14] | Normalized (/20 -> %): [60.0, 70.0, ...]</code>


```python
def stats_et_filtrage_simple(notes):
    """
    Simple Approach: Explicit loops and manual accumulators.
    Ideal for beginners to trace every step.
    """
    if not notes:
        return {"moyenne": 0, "mediane": 0, "superieures": [], "normalisees": []}
    
    # 1. Manual calculation of sum and mean
    total = 0
    for n in notes:
        total += n
    moyenne = total / len(notes)
    
    # 2. Filter grades strictly above the mean
    notes_sup = []
    for n in notes:
        if n > moyenne:
            notes_sup.append(n)
            
    # 3. Calculate median
    notes_triees = sorted(notes)
    n = len(notes_triees)
    if n % 2 == 1:
        mediane = notes_triees[n // 2]
    else:
        mediane = (notes_triees[n // 2 - 1] + notes_triees[n // 2]) / 2
        
    # 4. Normalize out of 100 (assuming grades out of 20)
    normalisees = []
    for n in notes_sup:
        normalisees.append((n / 20) * 100)
        
    return {
        "moyenne": round(moyenne, 2),
        "mediane": round(mediane, 2),
        "superieures": notes_sup,
        "normalisees": [round(x, 1) for x in normalisees]
    }


def stats_et_filtrage_optimal(notes):
    """
    Optimal / Pythonic Approach:
    - sum() / len() with guard clause
    - List comprehension for O(N) filtering and transformation
    - Concise O(N log N) median calculation
    """
    if not notes:
        return {"moyenne": 0, "mediane": 0, "superieures": [], "normalisees": []}
        
    moyenne = sum(notes) / len(notes)
    
    # Expressive single-line filtering
    notes_sup = [n for n in notes if n > moyenne]
    
    # Concise median calculation
    s = sorted(notes)
    mid = len(s) // 2
    mediane = s[mid] if len(s) % 2 != 0 else (s[mid - 1] + s[mid]) / 2
    
    # Vectorized normalization via list comprehension
    normalisees = [round((n / 20) * 100, 1) for n in notes_sup]
    
    return {
        "moyenne": round(moyenne, 2),
        "mediane": round(mediane, 2),
        "superieures": notes_sup,
        "normalisees": normalisees
    }


# --- Tests & Demonstration ---
notes_test = [12, 4, 14, 11, 18, 13, 7, 10, 5, 9, 15, 8, 14, 16]

res_simple = stats_et_filtrage_simple(notes_test)
res_optimal = stats_et_filtrage_optimal(notes_test)

print("=== Challenge 1: Statistics and Filtering ===")
print("Raw data:", notes_test)
print("Mean:", res_optimal["moyenne"])
print("Median:", res_optimal["mediane"])
print("Grades > Mean:", res_optimal["superieures"])
print("Normalized (out of 100):", res_optimal["normalisees"])
assert res_simple == res_optimal, "Both methods must return the same result!"
print("Empty list test:", stats_et_filtrage_optimal([]))
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **How It Works:**
   - **Mean vs. Median**: The mean is sensitive to extreme outliers, while the median represents the exact middle element that splits the distribution in half. For an even length $N$, the median is the average of elements at indices $\frac{N}{2}-1$ and $\frac{N}{2}$.
   - **List Comprehension vs. Standard Loop**: The syntax `[n for n in notes if n > moyenne]` executes at the CPython bytecode level, making it roughly 30% to 40% faster than an explicit `for` loop using `.append()`.

2. **Complexity:**
   - **Time Complexity**: $O(N \log N)$, dominated by `sorted(notes)` for the median. Calculating the mean and filtering are both $O(N)$.
   - **Space Complexity**: $O(N)$ to store the filtered lists and the sorted copy.

3. **Common Pitfalls to Avoid:**
   - **Division by Zero**: Never run `sum(notes)/len(notes)` without a guard clause `if not notes:`.
   - **Integer Division `//` vs. Float Division `/`**: Use `//` for integer index arithmetic (`n // 2`), but `/` for accurate floating-point averages.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 2: Word Intersection, Text Cleaning & Order Preservation</span>
<span style="background-color: #28a745; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Foundational</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Strings • Lists • Sets • Order Preservation</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Find common words between two text strings after stripping punctuation and converting to lowercase, while strictly preserving the word appearance order from the first string.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
A simple `set(t1) & set(t2)` finds common words in $O(N + M)$, but destroys word order. In text processing and search engines, preserving original document appearance order is critical.

* **Input Example**:  
  <code>t1 = "Python is powerful, elegant, and fast! Python rocks."</code>  
  <code>t2 = "Fast code in Python is always elegant and clean."</code>
* **Expected Output Example**:  
  <code>['python', 'is', 'elegant', 'and', 'fast']</code> (duplicates removed, order of t1 preserved)


```python
import string

def nettoyer_texte(texte):
    """Removes punctuation and normalizes to lowercase."""
    traducteur = str.maketrans("", "", string.punctuation)
    return texte.translate(traducteur).lower().split()


def intersection_mots_simple(texte1, texte2):
    """
    Simple Approach: Linear loop with 'in' membership checks.
    Order is preserved, but O(N * M) lookup complexity.
    """
    mots1 = nettoyer_texte(texte1)
    mots2 = nettoyer_texte(texte2)
    
    communs = []
    for m in mots1:
        if m in mots2 and m not in communs:
            communs.append(m)
            
    return communs


def intersection_mots_optimal(texte1, texte2):
    """
    Optimal Approach:
    - Pre-indexes words from text2 into a hash set for O(1) membership checks
    - Tracks seen elements with a second set in O(1) to avoid duplicates
    - Preserves order from text1 in total O(N + M) time
    """
    mots1 = nettoyer_texte(texte1)
    ensemble_mots2 = set(nettoyer_texte(texte2))
    
    deja_vus = set()
    communs = []
    
    for m in mots1:
        if m in ensemble_mots2 and m not in deja_vus:
            communs.append(m)
            deja_vus.add(m)
            
    return communs


# --- Tests & Demonstration ---
t1 = "Python is powerful, elegant, and fast! Python rocks."
t2 = "Fast code in Python is always elegant and clean."

res_simple = intersection_mots_simple(t1, t2)
res_optimal = intersection_mots_optimal(t1, t2)

print("=== Challenge 2: Word Intersection ===")
print("Text 1:", t1)
print("Text 2:", t2)
print("Common words (order preserved):", res_optimal)
assert res_simple == res_optimal
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Why `word in list` is a Performance Trap:**
   - In `intersection_mots_simple`, checking `m in mots2` scans the list item by item in $O(M)$. For $N$ words in `mots1`, the total cost is $O(N \times M)$.
   - Converting `mots2` to a `set` drops the lookup cost to $O(1)$ on average via hash tables. The total complexity drops to $O(N + M)$.

2. **Order Preservation vs. Raw Set Intersection:**
   - `list(set(mots1) & set(mots2))` is concise but loses original word order because hash tables do not guarantee input order across set operations.
   - Using a set for lookup (`ensemble_mots2`) while iterating over the original list (`mots1`) gives you the best of both worlds: $O(1)$ lookups with preserved order.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 3: Mixed Type Separation & The Boolean vs. Integer Trap</span>
<span style="background-color: #28a745; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Foundational</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Types • isinstance vs type • Listes • Data Cleaning</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Clean and categorize a heterogeneous list into 4 distinct sorted typed buckets: numbers (int/float), booleans, strings, and other objects, avoiding the classic `bool is a subclass of int` pitfall.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
In Python, `bool` inherits from `int` (`isinstance(True, int)` evaluates to `True`!). If you filter integers using `isinstance(x, (int, float))`, booleans like `True` and `False` will secretly pollute your numeric data.

* **Input Example**:  
  <code>stock = [42, "Laptop", True, 3.14, "Mouse", False, [1, 2], 0, -5, "Keyboard"]</code>
* **Expected Output Example**:  
  <code>Numbers: [-5, 0, 3.14, 42] | Booleans: [False, True] | Strings: ['Keyboard', 'Laptop', 'Mouse'] | Others: [[1, 2]]</code>


```python
def separer_et_trier_simple(stock):
    """
    Simple Approach:
    - Loop with manual type checks
    - Resolves the trap: 'bool' before 'int/float'!
    """
    chaines = []
    nombres = []
    booleens = []
    
    for element in stock:
        # IMPORTANT: In Python, isinstance(True, int) is True!
        # Must check bool first or use exact type(x) comparison.
        if isinstance(element, bool):
            booleens.append(element)
        elif isinstance(element, (int, float)):
            nombres.append(element)
        elif isinstance(element, str):
            chaines.append(element)
            
    # Sorting
    # str.lower enables case-insensitive alphabetical sorting
    chaines.sort(key=lambda s: s.lower())
    nombres.sort(reverse=True)
    
    return {
        "chaines": chaines,
        "nombres": nombres,
        "booleens": booleens
    }


def separer_et_trier_optimal(stock):
    """
    Optimal / Idiomatic Approach:
    - Uses exact type identity type(x) is bool
    - Clean list comprehensions
    - sorted() with custom key functions
    """
    # Precise filtering without subclass leakage
    booleens = [x for x in stock if type(x) is bool]
    nombres = sorted(
        [x for x in stock if (type(x) is int or type(x) is float)],
        reverse=True
    )
    chaines = sorted(
        [x for x in stock if type(x) is str],
        key=str.lower
    )
    
    return {
        "chaines": chaines,
        "nombres": nombres,
        "booleens": booleens
    }


# --- Tests & Demonstration ---
stock_mixte = ["Stylo", 25, "Classeur", 100.5, True, "Crayon", 12, False, "feutre", 5]

print("=== Challenge 3: Type Separation & The Boolean Trap ===")
print("Input list:", stock_mixte)

# Demonstrate the Python trap
print(f"PYTHON TRAP: isinstance(True, int) == {isinstance(True, int)}")
print(f"SOLUTION   : type(True) is int == {type(True) is int}")

res = separer_et_trier_optimal(stock_mixte)
print("\nResults after partitioning and sorting:")
print("  Strings (case-insensitive) :", res["chaines"])
print("  Numbers (descending)       :", res["nombres"])
print("  Booleans (cleanly isolated):", res["booleens"])

assert True not in res["nombres"], "Error: a boolean leaked into numbers!"
assert False not in res["nombres"], "Error: a boolean leaked into numbers!"
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **The Fundamental Trap: `bool` is an `int` in Python:**
   - In Python's object hierarchy: `issubclass(bool, int) == True`.
   - As a result: `isinstance(True, int)` returns `True`!
   - If you write `if isinstance(x, (int, float)): numbers.append(x)` before checking for `bool`, `True` and `False` will be treated as `1` and `0` and inserted into your numbers bucket.

2. **Resolution Rules:**
   - **Method 1**: Check `isinstance(x, bool)` strictly **before** `isinstance(x, (int, float))`.
   - **Method 2**: Use exact type equality `type(x) is int` if you do not want subclass inheritance to trigger.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 4: Linear Search vs. Binary Search in O(log N)</span>
<span style="background-color: #28a745; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Foundational</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Binary Search • Algorithmic Complexity • Two Pointers</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Implement both linear search $O(N)$ and binary search $O(\log N)$ on a sorted collection, comparing performance and iteration counts.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Finding an element with `.index()` or `in` performs an $O(N)$ scan. When data is already sorted, binary search cuts the remaining search space in half at every step.

* **Input Example**:  
  Sorted list of 1,000,000 elements, target element at index 750,000.
* **Expected Output Example**:  
  Linear: 750,001 comparisons. Binary: $\le 20$ comparisons.


```python
def recherche_lineaire(element, liste):
    """Linear scan in O(N). Returns (index, comparisons_count)."""
    comparaisons = 0
    for i in range(len(liste)):
        comparaisons += 1
        if liste[i] == element:
            return {"index": i, "comparaisons": comparaisons}
    return {"index": -1, "comparaisons": comparaisons}


def recherche_binaire(element, liste_triee):
    """
    Binary search in O(log N).
    Maintains low and high pointers, dividing the window by 2 each step.
    """
    gauche = 0
    droite = len(liste_triee) - 1
    comparaisons = 0
    
    while gauche <= droite:
        comparaisons += 1
        milieu = (gauche + droite) // 2
        valeur_milieu = liste_triee[milieu]
        
        if valeur_milieu == element:
            return {"index": milieu, "comparaisons": comparaisons}
        elif valeur_milieu < element:
            gauche = milieu + 1
        else:
            droite = milieu - 1
            
    return {"index": -1, "comparaisons": comparaisons}


# --- Tests & Demonstration ---
taille = 1_000_000
donnees_triees = list(range(taille))
cible = 789_456

res_lin = recherche_lineaire(cible, donnees_triees)
res_bin = recherche_binaire(cible, donnees_triees)

print("=== Challenge 4: Linear vs Binary Search ===")
print(f"Target: {cible} in a list of {taille:,} elements")
print(f"Linear: index = {res_lin['index']} in {res_lin['comparaisons']:,} steps")
print(f"Binary: index = {res_bin['index']} in {res_bin['comparaisons']} steps")

assert res_bin["index"] == res_lin["index"]
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **The Power of $O(\log N)$ vs. $O(N)$:**
   - In a dataset of 1,000,000 items, linear search requires up to 1,000,000 iterations.
   - Binary search requires at most $\log_2(1,000,000) \approx 20$ iterations!
   - In a dataset of 1,000,000,000 items, binary search needs only $\approx 30$ iterations.

2. **The Golden Rule:**
   - Binary search strictly requires the collection to be **sorted beforehand**.
   - If the list is unsorted and you only query it once, linear search $O(N)$ is faster than sorting ($O(N \log N)$) then searching.
   - If you query the same data multiple times, sorting once pays dividends immediately.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 5: Manual Occurrence Counting, Frequency Maps & Majority Element</span>
<span style="background-color: #28a745; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Foundational</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Lists • Hash Maps • Counter • Boyer-Moore Algorithm</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Count item frequencies manually without `liste.count()` in $O(N)$, build a complete frequency map, and find the majority element (appears $> N/2$ times) in $O(1)$ space using the Boyer-Moore Voting Algorithm.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Calling `liste.count(x)` inside a loop creates an $O(N^2)$ bottleneck. A single pass using a hash map solves this in $O(N)$. For finding a dominant element, Boyer-Moore solves it in $O(N)$ time and $O(1)$ auxiliary memory.

* **Input Example**:  
  <code>elements = [23, 7, 23, 19, 23, 12, 23, 7, 29, 23, 23]</code>
* **Expected Output Example**:  
  <code>Frequencies: {23: 6, 7: 2, 19: 1, 12: 1, 29: 1} | Majority Element: 23</code>


```python
from collections import Counter

def frequences_manuelles(liste):
    """Builds a frequency dictionary in a single O(N) pass."""
    frequences = {}
    for element in liste:
        frequences[element] = frequences.get(element, 0) + 1
    return frequences


def element_majoritaire_boyer_moore(liste):
    """
    Boyer-Moore Voting Algorithm:
    Finds the element that appears more than len(liste) // 2 times.
    Time Complexity: O(N)
    Space Complexity: O(1) auxiliary memory!
    """
    candidat = None
    vote = 0
    
    # 1. Candidate selection
    for x in liste:
        if vote == 0:
            candidat = x
            vote = 1
        elif x == candidat:
            vote += 1
        else:
            vote -= 1
            
    # 2. Verification step
    occurrences = sum(1 for x in liste if x == candidat)
    if occurrences > len(liste) // 2:
        return candidat
    return None


# --- Tests & Demonstration ---
L = [23, 7, 23, 19, 23, 12, 23, 7, 29, 23, 23]

freqs = frequences_manuelles(L)
majoritaire = element_majoritaire_boyer_moore(L)

print("=== Challenge 5: Frequencies and Majority Element ===")
print("Input list:", L)
print("Frequencies:", freqs)
print(f"Majority element (> {len(L)//2} times):", majoritaire)

assert freqs[23] == 6
assert freqs[7] == 2
assert majoritaire == 23
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Why You Should NEVER Do `for x in liste: c = liste.count(x)`:**
   - `.count()` scans the entire list in $O(N)$ on every single call.
   - Calling it inside a loop over $N$ items results in $O(N^2)$ time. On 100,000 items, this means 10 billion comparisons!
   - Building a frequency dictionary with `.get(x, 0) + 1` takes only $O(N)$ total time.

2. **The Elegance of Boyer-Moore:**
   - If an element appears more than half the time, its occurrences will cancel out all other elements combined.
   - This algorithm finds the dominant element in $O(1)$ memory, avoiding any dictionary allocations.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 6: Multi-Stage Functional Pipeline Engine (Filter -> Map -> Sorted)</span>
<span style="background-color: #fd7e14; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Advanced</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Functional Programming • Higher-Order Functions • Pipelines</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Build a reusable higher-order function that chains filtering, transformation (mapping), and sorting stages, supporting custom lambdas and lazy generator evaluation.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
In real-world data processing, transformations are chained: raw data -> clean -> filter -> transform -> sort. We encapsulate this flow in a composable engine that avoids intermediate memory allocations.

* **Input Example**:  
  <code>nombres = [12, 32, -2, 2, -21, 200, -65, 0, 15]</code>  
  Pipeline 1: keep positive numbers > 0, cube them ($x^3$), sort descending.
* **Expected Output Example**:  
  <code>[8000000, 32768, 3375, 1728, 8]</code>


```python
def pipeline_simple(donnees, critere_filtre, fonction_transfo, reverse=False):
    """
    Simple Approach: Step-by-step intermediate lists.
    Easy to trace and debug each stage.
    """
    # Step 1: Filtering
    filtres = []
    for item in donnees:
        if critere_filtre(item):
            filtres.append(item)
            
    # Step 2: Transformation
    transformes = []
    for item in filtres:
        transformes.append(fonction_transfo(item))
        
    # Step 3: Sorting
    transformes.sort(reverse=reverse)
    return transformes


def pipeline_optimal(donnees, critere_filtre=None, fonction_transfo=None, cle_tri=None, reverse=False):
    """
    Optimal / Functional Approach:
    - Uses generator expressions to avoid unnecessary intermediate list allocations in memory.
    - Single final allocation via sorted().
    """
    flux = donnees
    
    if critere_filtre is not None:
        flux = (x for x in flux if critere_filtre(x))
        
    if fonction_transfo is not None:
        flux = (fonction_transfo(x) for x in flux)
        
    if cle_tri is not None:
        return sorted(flux, key=cle_tri, reverse=reverse)
    return sorted(flux, reverse=reverse)


# --- Tests & Demonstration ---
nombres = [12, 32, -2, 2, -21, 200, -65, 0, 15]

# Pipeline 1: Positive numbers -> Cubed -> Sorted Descending
res_p1 = pipeline_optimal(
    nombres,
    critere_filtre=lambda x: x > 0,
    fonction_transfo=lambda x: x ** 3,
    reverse=True
)

# Pipeline 2: Even numbers -> Integer division by 2 -> Sorted Ascending
res_p2 = pipeline_optimal(
    nombres,
    critere_filtre=lambda x: x % 2 == 0,
    fonction_transfo=lambda x: x // 2,
    reverse=False
)

print("=== Challenge 6: Functional Pipeline ===")
print("Raw data:", nombres)
print("Pipeline 1 (Positives -> x^3 -> Descending):", res_p1)
print("Pipeline 2 (Evens -> x // 2 -> Ascending):", res_p2)

assert res_p1 == [8000000, 32768, 3375, 1728, 8]
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Generators `(x for x in ...)` vs. Lists `[x for x in ...]`:
   - An intermediate list `[fn(x) for x in data if cond(x)]` allocates a complete array in RAM before passing it to the next step.

2. **Higher-Order Functions in Modern Python:**
   - Functions in Python are first-class citizens: they can be passed as arguments, stored in variables, and returned by other functions.

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Real datasets frequently have ties that must be resolved cleanly across secondary and tertiary attributes. Python's lexicographical tuple sorting provides an elegant solution.

* **Input Example**:

```python
def tri_multi_criteres(etudiants):
    """
    Sorts students by:
    1. Grade descending (highest first)
    2. Age ascending (youngest first in case of tie)
    3. Name alphabetically ascending (tie breaker)
    
    KEY TECHNIQUE:
    Python compares tuples element by element: (a1, b1, c1) < (a2, b2, c2).
    To reverse a numeric criteria without setting global reverse=True,
    simply negate the number: -e['note']!
    """
    return sorted(
        etudiants,
        key=lambda e: (-e['note'], e['age'], e['nom'].lower())
    )


# --- Tests & Demonstration ---
etudiants = [
    {'nom': 'Alice', 'note': 85, 'age': 22},
    {'nom': 'Bob', 'note': 92, 'age': 20},
    {'nom': 'Charlie', 'note': 85, 'age': 19},
    {'nom': 'Diana', 'note': 92, 'age': 23},
    {'nom': 'Eve', 'note': 85, 'age': 22}
]

resultat_tri = tri_multi_criteres(etudiants)

print("=== Challenge 7: Multi-Criteria Sorting with Priority Tuples ===")
print("Final Ranking:")
for rank, e in enumerate(resultat_tri, 1):
    print(f"  {rank}. {e['nom']:<8} | Grade: {e['note']:<3} | Age: {e['age']}")

assert resultat_tri[0]['nom'] == 'Bob'
assert resultat_tri[1]['nom'] == 'Diana'
assert resultat_tri[2]['nom'] == 'Charlie'
assert resultat_tri[3]['nom'] == 'Alice'
assert resultat_tri[4]['nom'] == 'Eve'
print("All tie-breaking criteria validated successfully!")
```

<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 8: Single-Pass Dynamic Partitioning in O(N)</span>
<span style="background-color: #fd7e14; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Advanced</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Advanced Lists • Partitioning • Algorithmic Optimization</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Partition a list into two sublists (satisfied / not satisfied) in a single O(N) pass, without traversing the list twice with filter().

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
In basic exercises, solutions often use two successive filters: `filter(cond)` then `filter(not cond)`. This reads data twice. In streaming and large datasets, a single-pass partition is far superior.

* **Input Example**:  
  <code>nombres = [-5, 12, 0, -3, 8, 14, -1], critere = lambda x: x >= 0</code>
* **Expected Output Example**:  
  <code>Non-negative: [12, 0, 8, 14] | Negative: [-5, -3, -1]</code>


```python
def partitionner_deux_passes(liste, critere):
    """Naive Approach (Two passes): O(2N) list traversals."""
    satisfaits = [x for x in liste if critere(x)]
    non_satisfaits = [x for x in liste if not critere(x)]
    return satisfaits, non_satisfaits


def partitionner_une_passe_optimale(liste, critere):
    """
    Optimal Approach (Single pass): O(N).
    Each element is evaluated once and only once.
    """
    satisfaits = []
    non_satisfaits = []
    
    for element in liste:
        if critere(element):
            satisfaits.append(element)
        else:
            non_satisfaits.append(element)
            
    return satisfaits, non_satisfaits


def partitionner_multi_classes(liste, classes_dict):
    """
    Professional Extension: N-ary partitioning!
    Takes a dictionary of predicates: {'label': predicate_fn}
    """
    resultats = {label: [] for label in classes_dict}
    resultats["autres"] = []
    
    for element in liste:
        assigne = False
        for label, predicate in classes_dict.items():
            if predicate(element):
                resultats[label].append(element)
                assigne = True
                break
        if not assigne:
            resultats["autres"].append(element)
            
    return resultats


# --- Tests & Demonstration ---
nombres = [-5, 12, 0, -3, 8, 14, -1, 100]

pos, neg = partitionner_une_passe_optimale(nombres, lambda x: x >= 0)
print("=== Challenge 8: Single-Pass Partitioning ===")
print("Original list:", nombres)
print("Non-negative :", pos)
print("Negative     :", neg)

# Test multi-class partition
categories = {
    "mineurs": lambda x: x < 0,
    "petits": lambda x: 0 <= x < 10,
    "moyens": lambda x: 10 <= x < 50
}
part_multi = partitionner_multi_classes(nombres, categories)
print("\nMulti-class partitioning:")
for cat, valeurs in part_multi.items():
    print(f"  [{cat}]: {valeurs}")

assert pos == [12, 0, 8, 14, 100]
assert neg == [-5, -3, -1]
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Why 1 Single Pass is Superior to 2 Passes:**
   - On small in-memory lists, the wall-clock difference is negligible.
   - But on lists with millions of rows, or streams coming from a **network socket or file reader**, you cannot 'rewind' the stream without substantial caching cost.
   - Evaluating `critere(element)` only once also saves CPU cycles if the criterion is computationally heavy (such as regex matching or API validation).


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 9: Custom Accumulator & Reduction Engine with History (Scan & Fold)</span>
<span style="background-color: #fd7e14; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Advanced</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Functional Programming • reduce • Accumulator • Running Totals</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Implement a custom version of `functools.reduce` without importing any library, handling optional initial values, and building an accumulator that returns the full history of intermediate running states (Scan).

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Reduction (Fold) aggregates an iterable down to a single final value. Scan produces running totals (cumulative sums, moving minimums), a core tool in financial and time-series engineering.

* **Input Example**:  
  <code>data = [1, 2, 3, 4, 5], operation = lambda acc, x: acc + x</code>
* **Expected Output Example**:  
  <code>Reduce: 15 | Scan (Running history): [1, 3, 6, 10, 15]</code>


```python
def reduire_avance(iterable, operation, valeur_initiale=None):
    """
    Custom implementation of functools.reduce:
    - If initial value is provided, accumulator starts with it.
    - If not, accumulator takes the first element and iterates from the second.
    """
    it = iter(iterable)
    
    if valeur_initiale is None:
        try:
            accumulateur = next(it)
        except StopIteration:
            raise TypeError("reduire_avance() of empty iterable with no initial value")
    else:
        accumulateur = valeur_initiale
        
    for element in it:
        accumulateur = operation(accumulateur, element)
        
    return accumulateur


def scan_avance(iterable, operation, valeur_initiale=None):
    """
    Scan Operator (Cumulative Reduction):
    Returns a list containing all successive accumulated states.
    """
    it = iter(iterable)
    historique = []
    
    if valeur_initiale is None:
        try:
            acc = next(it)
            historique.append(acc)
        except StopIteration:
            return []
    else:
        acc = valeur_initiale
        historique.append(acc)
        
    for element in it:
        acc = operation(acc, element)
        historique.append(acc)
        
    return historique


# --- Tests & Demonstration ---
donnees = [1, 2, 3, 4, 5]

somme = reduire_avance(donnees, lambda a, b: a + b)
produit = reduire_avance(donnees, lambda a, b: a * b)
cumul_somme = scan_avance(donnees, lambda a, b: a + b)

print("=== Challenge 9: Custom Reduce and Scan ===")
print("Input data:", donnees)
print("Sum (reduce)     :", somme)
print("Product (reduce) :", produit)
print("Running sum (scan):", cumul_somme)

assert somme == 15
assert produit == 120
assert cumul_somme == [1, 3, 6, 10, 15]
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Fundamental Difference Between `reduce` (Fold) and `scan`:**
   - **`reduce`** collapses an entire collection into a single scalar value (e.g., total sum, global maximum, boolean AND).
   - **`scan`** (equivalent to `itertools.accumulate`) retains the entire trajectory of partial states. It is the mathematical foundation of cumulative metrics like account balances and moving averages.

2. **The `iter(iterable)` Pattern in Python:**
   - Calling `it = iter(iterable)` works universally on lists, tuples, sets, generators, and file objects.
   - `next(it)` fetches the first element cleanly, advancing the iterator so the subsequent `for x in it` loop automatically starts at the second element.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 10: Efficient Merging of Two Sorted Lists in O(N + M) without sort()</span>
<span style="background-color: #fd7e14; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Advanced</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Two Pointers • Merge Sort • Algorithmic Complexity</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Merge two pre-sorted lists into a single sorted list with duplicates removed in $O(N + M)$ time and without calling `.sort()` or `sorted()`.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Concatenating two sorted lists and calling `.sort()` costs $O((N + M) \log(N + M))$. The classic Two-Pointer algorithm exploits existing sorted order to merge in linear $O(N + M)$ time.

* **Input Example**:  
  <code>L1 = [1, 3, 5, 7, 9, 11]</code>  
  <code>L2 = [2, 3, 6, 8, 9, 10, 12, 14]</code>
* **Expected Output Example**:  
  <code>[1, 2, 3, 5, 6, 7, 8, 9, 10, 11, 12, 14]</code> (linear time, sorted, deduplicated)


```python
def fusion_deux_pointeurs_dedupliquee(l1, l2):
    """
    Two-Pointer Merge in O(N + M):
    Advances index pointers i and j based on value comparisons.
    Inline deduplication avoids duplicate elements.
    """
    i = 0
    j = 0
    n1 = len(l1)
    n2 = len(l2)
    fusion = []
    
    def ajouter(val):
        """Helper to append only if different from the last inserted item."""
        if not fusion or fusion[-1] != val:
            fusion.append(val)
            
    # Traverse while both pointers are in bounds
    while i < n1 and j < n2:
        if l1[i] < l2[j]:
            ajouter(l1[i])
            i += 1
        elif l2[j] < l1[i]:
            ajouter(l2[j])
            j += 1
        else:
            # Equal values: append once and advance both pointers
            ajouter(l1[i])
            i += 1
            j += 1
            
    # Append any remaining elements
    while i < n1:
        ajouter(l1[i])
        i += 1
        
    while j < n2:
        ajouter(l2[j])
        j += 1
        
    return fusion


# --- Tests & Demonstration ---
liste_a = [1, 3, 5, 7, 9, 11, 15]
liste_b = [2, 3, 6, 8, 9, 10, 12, 14, 15, 20]

resultat_fusion = fusion_deux_pointeurs_dedupliquee(liste_a, liste_b)
attendu = sorted(list(set(liste_a + liste_b)))

print("=== Challenge 10: Two-Pointer Merge ===")
print("List A :", liste_a)
print("List B :", liste_b)
print("Merged :", resultat_fusion)

assert resultat_fusion == attendu, f"Error: Got {resultat_fusion} vs {attendu}"
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Why This Technique is a Technical Interview Classic:**
   - It is the core subroutine of **Merge Sort** ($O(N \log N)$).
   - It demonstrates that you recognize when data is already ordered: you avoid throwing away existing structure with a brute force `.sort()`.
   - Linear complexity $O(N + M)$ is mathematically optimal, because every element must be inspected at least once.

2. **In-Memory Merging of Gigantic Datasets:**
   - In big data and database engines, this same algorithm drives the **Sort-Merge Join** between two massive disk-backed tables that cannot fit into memory simultaneously.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 11: Dictionary Partitioning & Safe Key/Value Inversion</span>
<span style="background-color: #28a745; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Foundational</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Dictionaries • Partitioning • Inversion • Collision Handling</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Partition a dictionary into two sub-dictionaries using a predicate, then implement safe key-value inversion that preserves all original keys even when multiple keys share identical values.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Naive inversion `{v: k for k, v in d.items()}` silently overwrites keys whenever values repeat. Safe inversion groups all original keys under a list or set bucket: `{value: [key1, key2]}`.

* **Input Example**:  
  <code>notes = {"Amine": 15.5, "Yassine": 19.0, "Sara": 15.5, "Malak": 8.7}</code>
* **Expected Output Example**:  
  <code>Inversion: {15.5: ['Amine', 'Sara'], 19.0: ['Yassine'], 8.7: ['Malak']}</code>


```python
def partitionner_dictionnaire(dico, predicat):
    """
    Partitions a dictionary into two sub-dictionaries (True / False)
    based on a predicate applied to each (key, value) pair.
    """
    vrais = {}
    faux = {}
    
    for cle, valeur in dico.items():
        if predicat(cle, valeur):
            vrais[cle] = valeur
        else:
            faux[cle] = valeur
            
    return vrais, faux


def inverser_dictionnaire_securise(dico):
    """
    Safely inverts keys and values:
    Groups conflicting keys under a list: {value: [key1, key2]}
    """
    inverse = {}
    for cle, valeur in dico.items():
        if valeur not in inverse:
            inverse[valeur] = []
        inverse[valeur].append(cle)
    return inverse


# --- Tests & Demonstration ---
notes_classe = {
    "Amine": 15.5,
    "Yassine": 19.0,
    "Sara": 15.5,
    "Malak": 8.7,
    "Karim": 10.0,
    "Hiba": 8.7
}

admis, ajournes = partitionner_dictionnaire(notes_classe, lambda nom, note: note >= 10)
dico_inverse = inverser_dictionnaire_securise(notes_classe)

print("=== Challenge 11: Dictionary Partitioning and Inversion ===")
print("Original grades :", notes_classe)
print("Passing (>= 10) :", admis)
print("Failing (< 10)  :", ajournes)
print("\nSafely Inverted (Collision-Free):", dico_inverse)

assert dico_inverse[15.5] == ["Amine", "Sara"]
assert dico_inverse[8.7] == ["Malak", "Hiba"]
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **The Danger of Naive Inversion `{v: k for k, v in d.items()}`:**
   - In a dictionary, keys must be strictly unique.
   - If `Amine` has grade `15.5` and `Sara` has `15.5`, naive inversion silently overwrites `Amine` when processing `Sara`.
   - Safe inversion must always collect conflicting keys into a list or set: `{value: [key1, key2]}`.

2. **Hashability of Values:**
   - You can only invert a dictionary if its **values are hashable** (immutable objects like strings, numbers, or tuples).
   - If a dictionary value is a list or dictionary, attempting to use it as a dictionary key will immediately raise a `TypeError: unhashable type`.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 12: The 4 Modern Methods to Merge Dictionaries in Python</span>
<span style="background-color: #17a2b8; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Intermediate</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Dictionaries • Merging • Pipe Operator • Side Effects • Python 3.9+</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Master and compare the 4 official ways to merge dictionaries in Python: in-place `.update()`, dictionary unpacking `**`, modern union operator `|`, and zero-copy `collections.ChainMap`.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Combining configurations, state objects, and parameter maps is ubiquitous. Knowing memory consumption and mutation side effects across these 4 methods is essential.

* **Input Example**:  
  <code>d1 = {'a': 1, 'b': 2}, d2 = {'b': 99, 'c': 3}, d3 = {'d': 4}</code>
* **Expected Output Example**:  
  Merged dictionary where later keys override earlier ones: <code>{'a': 1, 'b': 99, 'c': 3, 'd': 4}</code>


```python
from collections import ChainMap

def demo_fusions_dictionnaires():
    d1 = {"a": 1, "b": 2}
    d2 = {"b": 99, "c": 3}
    d3 = {"d": 4}
    
    # Method 1: dict.update() (MODIFIES IN-PLACE the target object)
    copie_d1 = d1.copy()
    copie_d1.update(d2)
    copie_d1.update(d3)
    
    # Method 2: Dictionary unpacking ** (Python 3.5+) - Creates a NEW dictionary
    fusion_unpack = {**d1, **d2, **d3}
    
    # Method 3: Union operator | (Python 3.9+) - The modern standard!
    fusion_pipe = d1 | d2 | d3
    
    # Method 4: collections.ChainMap - Zero-copy virtual view
    vue_chaine = ChainMap(d3, d2, d1)
    
    return {
        "update": copie_d1,
        "unpack": fusion_unpack,
        "pipe": fusion_pipe,
        "chainmap": vue_chaine
    }


# --- Tests & Demonstration ---
res = demo_fusions_dictionnaires()

print("=== Challenge 12: The 4 Merge Methods ===")
print("1. dict.update()     :", res["update"])
print("2. Unpack {**a, **b} :", res["unpack"])
print("3. Union operator |  :", res["pipe"])
print("4. ChainMap virtual  :", dict(res["chainmap"]))

assert res["update"] == res["unpack"] == res["pipe"]
assert res["pipe"] == {"a": 1, "b": 99, "c": 3, "d": 4}
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Comparison of Merge Methods:**

| Method | Python Version | Modifies Original? | Creates New Copy? | Best Use Case |
|---|---|---|---|---|
| `d1.update(d2)` | Python 2.x+ | **Yes (in-place)** | No (mutates `d1`) | Updating internal state in a class |
| `{**d1, **d2}` | Python 3.5+ | No (returns new) | Yes | Legacy compatibility (Python 3.6 - 3.8) |
| `d1 \| d2` | **Python 3.9+** | No (returns new) | Yes | Clean, expressive modern Python |
| `ChainMap(d2, d1)` | Python 3.3+ | No | **No (Zero-copy)** | Layered configurations (CLI > Env > Defaults) |

2. **The Power of `ChainMap`:**
   - `ChainMap` creates a live layered view without copying elements or allocating a new dictionary in memory.
   - Lookups search dictionaries in order from left to right: first match wins.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 13: Converting Lists to a Dictionary & Handling Unequal Lengths</span>
<span style="background-color: #17a2b8; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Intermediate</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Dictionaries • zip • zip_longest • Missing Data</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Map parallel lists into a dictionary using 3 distinct behaviors: truncation (standard `zip`), default padding (`zip_longest`), and strict validation (`strict=True`).

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
When header rows and data rows do not align in length, standard `zip` silently truncates excess elements. In data engineering, unhandled silent truncation causes serious data loss.

* **Input Example**:  
  <code>keys = ['name', 'age', 'city', 'country', 'role']</code>  
  <code>values = ['Alice', 30, 'Paris']</code>
* **Expected Output Example**:  
  <code>Truncate: 3 items | Fill: 5 items with missing values padded as 'Unknown'</code>


```python
from itertools import zip_longest

def conversion_listes_vers_dict(cles, valeurs, mode="strict", valeur_defaut="N/A"):
    """
    Converts two lists to a dictionary across 3 modes:
    - 'tronque': Stops at the shortest list (standard zip).
    - 'rempli' : Fills missing values with valeur_defaut (zip_longest).
    - 'strict' : Raises ValueError if lengths differ (Python 3.10+).
    """
    if mode == "tronque":
        return dict(zip(cles, valeurs))
        
    elif mode == "rempli":
        return dict(zip_longest(cles, valeurs, fillvalue=valeur_defaut))
        
    elif mode == "strict":
        try:
            return dict(zip(cles, valeurs, strict=True))
        except ValueError as e:
            raise ValueError(f"Lengths differ: {len(cles)} keys vs {len(valeurs)} values") from e
            
    else:
        raise ValueError(f"Unknown mode: '{mode}'")


# --- Tests & Demonstration ---
cles = ["nom", "age", "ville", "pays", "role"]
valeurs = ["Alice", 30, "Paris"]

dict_tronque = conversion_listes_vers_dict(cles, valeurs, mode="tronque")
dict_rempli = conversion_listes_vers_dict(cles, valeurs, mode="rempli", valeur_defaut="Inconnu")

print("=== Challenge 13: Converting Lists to Dictionary ===")
print("Keys   :", cles)
print("Values :", valeurs)
print("\n1. Truncated Mode :", dict_tronque)
print("2. Filled Mode    :", dict_rempli)

# Verify strict mode raises exception
try:
    conversion_listes_vers_dict(cles, valeurs, mode="strict")
except ValueError as e:
    print("3. Strict Mode    : Exception caught as expected ->", e)

assert len(dict_tronque) == 3
assert len(dict_rempli) == 5
assert dict_rempli["pays"] == "Inconnu"
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **The Silent Trap of `zip()` in Data Engineering:**
   - By default, `zip()` silently terminates as soon as the shorter iterable is exhausted.
   - If a source schema contains 10 columns but a malformed row has only 8, `zip()` will drop the final 2 columns without any warning or error.
   - In Python 3.10+, `zip(..., strict=True)` was explicitly introduced to prevent this bug by raising an immediate exception.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 14: Multi-Level Sorting & Inverting a Complex Inventory Dictionary</span>
<span style="background-color: #17a2b8; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Intermediate</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Dictionaries • sorted • Lambda Sorting • Transformations</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Sort a nested inventory dictionary on dynamic computed attributes (such as `unit_price * quantity`), handle descending orders, and restructure output into a ranking table.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Inventories are nested dictionaries: `{product_name: {'price': p, 'qty': q}}`. Sorting them requires extracting values from `(key, subdict)` tuples using lambda expressions.

* **Input Example**:  
  <code>inventory = {'Laptop': {'prix': 1200, 'qte': 5}, 'Souris': {'prix': 25, 'qte': 40}, 'Écran': {'prix': 300, 'qte': 10}}</code>
* **Expected Output Example**:  
  Sorted by total stock value: Laptop ($6000), Écran ($3000), Souris ($1000).


```python
def trier_inventaire(inventaire, critere="valeur_totale", reverse=True):
    """
    Sorts an inventory dictionary by:
    - 'prix': Unit price
    - 'qte': Available stock
    - 'valeur_totale': price * quantity
    - 'nom': Alphabetical product name
    """
    if critere == "prix":
        cle_func = lambda item: item[1]["prix"]
    elif critere == "qte":
        cle_func = lambda item: item[1]["qte"]
    elif critere == "valeur_totale":
        cle_func = lambda item: item[1]["prix"] * item[1]["qte"]
    elif critere == "nom":
        cle_func = lambda item: item[0].lower()
    else:
        raise ValueError(f"Unknown sorting criteria: '{critere}'")
        
    elements_tries = sorted(inventaire.items(), key=cle_func, reverse=reverse)
    return dict(elements_tries)


# --- Tests & Demonstration ---
inventaire = {
    "Laptop":  {"prix": 1200, "qte": 5},
    "Souris":  {"prix": 25,   "qte": 40},
    "Écran":   {"prix": 300,  "qte": 10},
    "Clavier": {"prix": 80,   "qte": 15},
    "Casque":  {"prix": 150,  "qte": 8}
}

tri_valeur = trier_inventaire(inventaire, critere="valeur_totale", reverse=True)

print("=== Challenge 14: Nested Inventory Sorting ===")
print(f"{'Product':<10} | {'Price':<6} | {'Qty':<4} | {'Total Value':<10}")
print("-" * 40)
for produit, info in tri_valeur.items():
    val = info["prix"] * info["qte"]
    print(f"{produit:<10} | {info['prix']:<6} | {info['qte']:<4} | ${val:<10}")

assert list(tri_valeur.keys())[0] == "Laptop"   # 1200 * 5 = 6000
assert list(tri_valeur.keys())[1] == "Écran"    # 300 * 10 = 3000
assert list(tri_valeur.keys())[2] == "Clavier"  # 80 * 15 = 1200
assert list(tri_valeur.keys())[3] == "Casque"   # 150 * 8 = 1200
assert list(tri_valeur.keys())[4] == "Souris"   # 25 * 40 = 1000
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **`item[0]` vs. `item[1]`:**
   - When calling `sorted(d.items(), key=...)`, each element passed to the lambda is a `(key, value)` tuple.
   - `item[0]` is the dictionary key (the product name).
   - `item[1]` is the associated value (here a sub-dictionary `{'prix': ..., 'qte': ...}`).
   - Writing `item[1]['prix'] * item[1]['qte']` calculates the metric on the fly without mutating the original dictionary structure.

2. **Insertion Order Guaranteed in Python 3.7+:**
   - Converting the sorted list of tuples back into a dictionary via `dict(...)` retains the sorted order reliably across iterations.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 15: Immutability, Hashability & Tuples as Spatial Dictionary Keys</span>
<span style="background-color: #fd7e14; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Advanced</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Tuples • Immutability • Hash Tables • Sparse Matrices</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Understand Python's hash contract, demonstrate why lists cannot serve as dictionary keys while tuples can, and build a memory-efficient sparse matrix using coordinate tuples `(x, y)` as keys.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
A 10,000 x 10,000 matrix contains 100,000,000 cells. If only 500 cells hold non-zero values, allocating a full 2D nested list wastes gigabytes of memory. A sparse dictionary indexed by coordinate tuples `(row, col)` stores only populated cells in $O(1)$ lookup time.

* **Input Example**:  
  Coordinates (10, 20) with value 42, coordinates (500, 300) with value 99 in a 1,000,000-cell grid.
* **Expected Output Example**:  
  Grid lookup (10, 20) -> 42. Grid lookup (1, 1) -> default 0. Memory footprint < 1 KB.


```python
def demo_hachabilite_et_tuples():
    """Demonstrates Python's hashing contract."""
    
    # 1. Standard immutable tuple is hashable
    tuple_cle = ("France", "Paris")
    print(f"Hash of tuple {tuple_cle}: {hash(tuple_cle)}")
    
    # 2. A list is mutable -> Unhashable
    try:
        liste_cle = ["France", "Paris"]
        {liste_cle: "Capital"}
    except TypeError as e:
        print("Caught expected error for list key:", e)
        
    # 3. SUBTLE TRAP: A tuple containing a mutable object (like a list) is NOT hashable!
    try:
        tuple_avec_liste = ("Paris", [1, 2, 3])
        {tuple_avec_liste: "Data"}
    except TypeError as e:
        print("Caught expected error for tuple containing list:", e)


class MatriceCreuse:
    """
    Sparse Matrix implementation using coordinate tuples (x, y) as keys.
    Memory consumption is proportional only to non-zero elements: O(K), where K << N * M.
    """
    def __init__(self, valeur_defaut=0):
        self.grille = {}
        self.valeur_defaut = valeur_defaut
        
    def set(self, x, y, valeur):
        if valeur == self.valeur_defaut:
            self.grille.pop((x, y), None)
        else:
            self.grille[(x, y)] = valeur
            
    def get(self, x, y):
        return self.grille.get((x, y), self.valeur_defaut)
        
    def __repr__(self):
        return f"MatriceCreuse({len(self.grille)} non-zero populated cells)"


# --- Tests & Demonstration ---
print("=== Challenge 15: Hashability and Sparse Matrix ===")
demo_hachabilite_et_tuples()

# Simulate a massive 10,000 x 10,000 grid
grille = MatriceCreuse(valeur_defaut=0)
grille.set(10, 20, 42)
grille.set(5000, 8000, 99)

print("\nMatrix status:", grille)
print("Lookup populated (10, 20):", grille.get(10, 20))
print("Lookup empty (1, 1)      :", grille.get(1, 1))

assert grille.get(10, 20) == 42
assert grille.get(1, 1) == 0
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **What is a Hashable Object?**
   - An object is hashable if it has an immutable hash value throughout its lifetime (`__hash__()`) and can be compared for equality with other objects (`__eq__()`).
   - Dictionary keys and set elements **must be hashable** because Python uses this integer hash to compute the bucket slot in the hash table in $O(1)$ time.

2. **The Tuple with Inner List Trap:**
   - A tuple is only hashable if **all elements inside it are also hashable**.
   - `("A", (1, 2))` is hashable.
   - `("A", [1, 2])` raises `TypeError: unhashable type: 'list'` because the inner list's contents can change, which would corrupt the hash bucket address.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 16: Multi-Dictionary Merge with Dynamic Conflict Resolution</span>
<span style="background-color: #fd7e14; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Advanced</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Dictionaries • Conflict Resolution • MapReduce • Custom Merges</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Merge $N$ dictionaries where duplicate keys are resolved dynamically using custom conflict resolution functions (such as sum, max, or list concatenation) rather than simply overwriting previous values.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Standard merges (`d1 | d2` or `.update()`) silently overwrite duplicate keys with the latest value. In distributed data systems (like MapReduce or metrics aggregation), colliding keys must be reduced using business logic (e.g., adding counters together).

* **Input Example**:  
  <code>d1 = {'a': 10, 'b': 20}, d2 = {'b': 35, 'c': 5}, d3 = {'b': 10, 'c': 15}</code>
* **Expected Output Example**:  
  Sum merge: `{'a': 10, 'b': 65, 'c': 20}` | Max merge: `{'a': 10, 'b': 35, 'c': 15}`


```python
def fusionner_n_dictionnaires(liste_dicts, regle_conflit=lambda ancien, nouveau: ancien + nouveau):
    """
    Merges N dictionaries sequentially with a customizable conflict resolution function.
    - If a key is seen for the first time, it is set directly.
    - If a key already exists, regle_conflit(existing_val, incoming_val) is called.
    """
    resultat = {}
    
    for dico in liste_dicts:
        for cle, valeur in dico.items():
            if cle in resultat:
                resultat[cle] = regle_conflit(resultat[cle], valeur)
            else:
                resultat[cle] = valeur
                
    return resultat


# --- Tests & Demonstration ---
d1 = {"a": 10, "b": 20}
d2 = {"b": 35, "c": 5}
d3 = {"b": 10, "c": 15, "d": 50}

# Merge 1: Accumulate sums
fusion_somme = fusionner_n_dictionnaires([d1, d2, d3], regle_conflit=lambda a, b: a + b)

# Merge 2: Keep maximum
fusion_max = fusionner_n_dictionnaires([d1, d2, d3], regle_conflit=lambda a, b: max(a, b))

# Merge 3: Collect values into a list
def collecter(ancien, nouveau):
    if isinstance(ancien, list):
        return ancien + [nouveau]
    return [ancien, nouveau]

fusion_liste = fusionner_n_dictionnaires([d1, d2, d3], regle_conflit=collecter)

print("=== Challenge 16: Multi-Dict Conflict Resolution ===")
print("Input dicts:", [d1, d2, d3])
print("Sum merge   :", fusion_somme)
print("Max merge   :", fusion_max)
print("List merge  :", fusion_liste)

assert fusion_somme["b"] == 20 + 35 + 10  # 65
assert fusion_max["b"] == 35
assert fusion_liste["b"] == [20, 35, 10]
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Why This Architecture is Central to Big Data / MapReduce:**
   - In distributed data systems (Hadoop, PySpark), the **Reduce** phase groups intermediate values from multiple worker nodes by key.
   - Passing a custom `regle_conflit` gives you a flexible, generalized aggregation engine for sums, moving averages, union sets, and audit logs.

<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Group By • defaultdict • Data Aggregation • Search Index</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Implement SQL-style `GROUP BY` using `defaultdict(list)`, compute aggregate statistics per group, and build a full-text inverted search index mapping tokens to document IDs.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Grouping tabular records by categorical attributes (department, city, category) is the most common analytical operation. `collections.defaultdict` provides an optimal, clean implementation pattern.

* **Input Example**:  
  Transactions: `[{'ville': 'Paris', 'montant': 120}, {'ville': 'Lyon', 'montant': 45}, {'ville': 'Paris', 'montant': 300}, ...]`
* **Expected Output Example**:  
  `Paris: sum = 470, count = 3 | Lyon: sum = 195, count = 2`


```python
from collections import defaultdict

def regrouper_avec_setdefault(elements, fonction_cle):
    """
    Approach 1: Idiomatic use of d.setdefault(key, []).append(...)
    Advantage: Requires no external module imports.
    """
    groupes = {}
    for item in elements:
        cle = fonction_cle(item)
        groupes.setdefault(cle, []).append(item)
    return groupes


def regrouper_avec_defaultdict(elements, fonction_cle):
    """
    Approach 2: Using collections.defaultdict(list)
    Advantage: Faster in CPython (avoids creating temporary list on existing keys).
    """
    groupes = defaultdict(list)
    for item in elements:
        groupes[fonction_cle(item)].append(item)
    return dict(groupes)


# --- Tests & Demonstration ---
transactions = [
    {"ville": "Paris", "montant": 120, "categorie": "Tech"},
    {"ville": "Lyon", "montant": 45, "categorie": "Food"},
    {"ville": "Paris", "montant": 300, "categorie": "Voyage"},
    {"ville": "Marseille", "montant": 80, "categorie": "Food"},
    {"ville": "Lyon", "montant": 150, "categorie": "Tech"},
    {"ville": "Paris", "montant": 50, "categorie": "Food"}
]

# Group by City using setdefault
par_ville = regrouper_avec_setdefault(transactions, lambda t: t["ville"])

# Group by Category using defaultdict
par_cat = regrouper_avec_defaultdict(transactions, lambda t: t["categorie"])

# Aggregation: Sum of amounts by city
somme_par_ville = {
    ville: sum(tx["montant"] for tx in txs)
    for ville, txs in par_ville.items()
}

print("=== Challenge 17: Hierarchical Grouping (Group By) ===")
print("Grouping by City:")
for ville, txs in par_ville.items():
    print(f"  [{ville}]: {len(txs)} transactions -> amounts: {[t['montant'] for t in txs]}")

print("\nSQL Aggregation SUM(montant) GROUP BY ville:")
for ville, total in somme_par_ville.items():
    print(f"  {ville:<10}: ${total}")

assert somme_par_ville["Paris"] == 470
assert somme_par_ville["Lyon"] == 195
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **The Hidden Pitfall of `dict.setdefault()`:**
   - Writing `d.setdefault(key, [])` allocates a new empty list `[]` in memory on **every invocation**, even if `key` already exists!
   - `collections.defaultdict(list)` avoids this waste: its factory function is called strictly when a key is truly missing.

2. **Inverted Index in Modern Search Engines:**
   - Inverted indices power search systems like Elasticsearch and Lucene.
   - Using sets for document IDs ensures instant $O(1)$ deduplication and supports fast boolean search queries via set intersections (`set1 & set2`).


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 18: Vector Transformation & Normalization of Numerical Sequences</span>
<span style="background-color: #17a2b8; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Intermediate</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Tuples • Comprehensions • Vector Transformations • Min-Max</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Transform collections of numerical tuples using list and tuple comprehensions, and implement coordinate Min-Max normalization to scale features between 0.0 and 1.0.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
In Machine Learning, coordinate vectors `(x, y)` often have mismatched scales. Normalizing them across dimensions without external libraries demonstrates clean sequence comprehension design.

* **Input Example**:  
  <code>vectors = [(10, 20), (30, 40), (50, 60)]</code>
* **Expected Output Example**:  
  Doubled: `[(20, 40), (60, 80), (100, 120)]` | Min-Max Normalized: `[(0.0, 0.0), (0.5, 0.5), (1.0, 1.0)]`


```python
def transformer_tuples_comprehension(liste_tuples, fonction):
    """
    Pythonic Approach:
    Nested comprehension (outer list comprehension, inner tuple constructor).
    Maximum readability and optimal execution speed.
    """
    return [
        tuple(fonction(val) for val in t)
        for t in liste_tuples
    ]


def normalisation_min_max_tuples(liste_tuples):
    """
    Real-World Machine Learning Task:
    Min-Max feature scaling of coordinate vectors between 0.0 and 1.0.
    """
    if not liste_tuples:
        return []
        
    dim = len(liste_tuples[0])
    
    # 1. Compute min and max along each dimension
    min_vals = [min(t[d] for t in liste_tuples) for d in range(dim)]
    max_vals = [max(t[d] for t in liste_tuples) for d in range(dim)]
    
    # 2. Normalize each coordinate: (x - min) / (max - min)
    normalises = []
    for t in liste_tuples:
        point_norm = []
        for d in range(dim):
            etendue = max_vals[d] - min_vals[d]
            val = (t[d] - min_vals[d]) / etendue if etendue != 0 else 0.0
            point_norm.append(round(val, 2))
        normalises.append(tuple(point_norm))
        
    return normalises


# --- Tests & Demonstration ---
coordonnees = [(10, 20), (30, 40), (50, 60)]

double = transformer_tuples_comprehension(coordonnees, lambda x: x * 2)
norm = normalisation_min_max_tuples(coordonnees)

print("=== Challenge 18: Vector Transformations ===")
print("Original coords:", coordonnees)
print("Doubled vectors:", double)
print("Min-Max scaled :", norm)

assert double == [(20, 40), (60, 80), (100, 120)]
assert norm == [(0.0, 0.0), (0.5, 0.5), (1.0, 1.0)]
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Why `tuple(...)` is Required in the Comprehension:**
   - Writing parentheses `(fonction(val) for val in t)` creates a **lazy generator**, not a tuple!
   - To produce an actual tuple object, you must explicitly wrap the generator expression with the `tuple(...)` constructor.

<span style="background-color: #fd7e14; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Advanced</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Advanced Dictionaries • Recursion • Flatten • JSON Parsing</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Recursively flatten deeply nested dictionaries of arbitrary depth into flat key-value pairs using dotted compound keys (`parent.child.subchild`).

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
JSON payloads from REST APIs frequently have deeply nested hierarchies. Relational databases require flat columnar tables. Flattening transforms hierarchical trees into relational rows.

* **Input Example**:  
  <code>D = {'a': 1, 'b': {'c': 2, 'd': {'e': 3}}}</code>
* **Expected Output Example**:  
  <code>{'a': 1, 'b.c': 2, 'b.d.e': 3}</code>


```python
def aplatir_dictionnaire_recursif(dico, prefixe="", separateur="."):
    """
    Recursively flattens a nested dictionary of arbitrary depth.
    Returns a list of tuples (compound_key, terminal_value).
    """
    elements = []
    
    for cle, valeur in dico.items():
        nouvelle_cle = f"{prefixe}{separateur}{cle}" if prefixe else str(cle)
        
        # Strict check: if value is a nested dict, recurse deeper
        if isinstance(valeur, dict):
            elements.extend(aplatir_dictionnaire_recursif(valeur, nouvelle_cle, separateur=separateur))
        else:
            elements.append((nouvelle_cle, valeur))
            
    return elements


# --- Tests & Demonstration ---
config_complexe = {
    "source": {
        "type": "postgres",
        "credentials": {
            "hote": "localhost",
            "port": 5432,
            "ssl": True
        }
    },
    "destination": {
        "cloud": "AWS",
        "s3": {
            "bucket": "data-lake-prod"
        }
    },
    "version": "2.1.0"
}

tuples_plats = aplatir_dictionnaire_recursif(config_complexe)
dict_plat = dict(tuples_plats)

print("=== Challenge 19: Recursive Flattening ===")
for cle, val in dict_plat.items():
    print(f"  {cle:<30} -> {val}")

assert dict_plat["source.credentials.port"] == 5432
assert dict_plat["destination.cloud"] == "AWS"
assert len(tuples_plats) == 7
print("Test passed successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Bug Analysis in Legacy Code:**
   - A common beginner mistake is checking `if not isinstance(v, tuple):` instead of `if isinstance(v, dict):`.
   - Always check `isinstance(valeur, dict)` explicitly when deciding whether to recurse down tree branches.

2. **Complexity:**
   - **Time Complexity**: $O(N)$ where $N$ is the total count of keys in the tree.
   - **Space Complexity**: $O(D)$ recursion stack depth, where $D$ is the maximum nesting level.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 20: Recursive Unflattening: Rebuilding the Nested Tree Structure</span>
<span style="background-color: #dc3545; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Boss Challenge</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Advanced Dictionaries • Tree Structures • Unflatten • Pointer Traversal</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Implement the exact mathematical inverse of Challenge 19: reconstruct an arbitrary nested dictionary tree from flat dotted keys.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
In config management and database serialization, you frequently receive flat environment variable paths (e.g. `APP.DATABASE.PORT=5432`) that must be converted into structured JSON trees.

* **Input Example**:  
  <code>{'user.profile.name': 'Alice', 'user.profile.age': 30, 'system.cpu.cores': 8}</code>
* **Expected Output Example**:  
  <code>{'user': {'profile': {'name': 'Alice', 'age': 30}}, 'system': {'cpu': {'cores': 8}}}</code>


```python
def desaplatir_dictionnaire(dico_plat, separateur="."):
    """
    Reconstructs a nested dictionary hierarchy from flat dotted keys.
    Uses reference pointer traversal to navigate or create intermediate branches.
    """
    racine = {}
    
    for cle_composite, valeur in dico_plat.items():
        parties = cle_composite.split(separateur)
        noeud_courant = racine
        
        # Traverse all segments except the last one to build/find branches
        for partie in parties[:-1]:
            if partie not in noeud_courant:
                noeud_courant[partie] = {}
            elif not isinstance(noeud_courant[partie], dict):
                # Path collision: a scalar already exists at this key
                noeud_courant[partie] = {"_valeur_precedente": noeud_courant[partie]}
                
            noeud_courant = noeud_courant[partie]
            
        # Assign the terminal value at the leaf key
        noeud_courant[parties[-1]] = valeur
        
    return racine


# --- Tests & Demonstration ---
dico_plat_test = {
    "user.profile.nom": "Hamza",
    "user.profile.age": 25,
    "user.localisation.pays": "Maroc",
    "user.localisation.ville": "Ouled_Teima",
    "systeme.cpu.coeurs": 8,
    "systeme.os": "Linux"
}

arbre_reconstruit = desaplatir_dictionnaire(dico_plat_test)

print("=== Challenge 20: Unflattening (Tree Reconstruction) ===")
import pprint
pprint.pprint(arbre_reconstruit)

assert arbre_reconstruit["user"]["profile"]["nom"] == "Hamza"
assert arbre_reconstruit["user"]["localisation"]["pays"] == "Maroc"
assert arbre_reconstruit["systeme"]["cpu"]["coeurs"] == 8
print("Tree successfully reconstructed and validated!")
```

#### Deep Analysis & Key Takeaways

1. **How the `noeud_courant` Navigation Pointer Works:**
   - In Python, dictionaries are passed and assigned **by reference**.
   - When executing `noeud_courant = noeud_courant[partie]`, the variable directly points to the sub-dictionary inside `racine`. Any future modifications through `noeud_courant` mutate the root tree in place!
   - This pointer traversal allows constructing arbitrary trees in $O(N \times K)$ time where $K$ is path depth.

<span style="background-color: #dc3545; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Boss Challenge</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Tuples • Hash Tables • Algorithms • LeetCode Medium</span>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Group a collection of words into anagram buckets using sorted character tuple signatures in optimal time and space.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
Two words are anagrams if they share identical characters in different orders. A list of sorted characters cannot be a dictionary key because lists are unhashable. Converting the sorted letters to a `tuple` makes it hashable.

* **Input Example**:  
  <code>words = ["listen", "silent", "enlist", "rat", "tar", "art", "evil", "vile", "live"]</code>
* **Expected Output Example**:  
  <code>[['listen', 'silent', 'enlist'], ['rat', 'tar', 'art'], ['evil', 'vile', 'live']]</code>


```python
def regrouper_anagrammes_tri(liste_mots):
    """
    Approach 1: Sorted Character Signature (Tuple).
    Two anagrams will have identical sorted tuples of characters.
    Ex: 'listen' -> ('e', 'i', 'l', 'n', 's', 't')
        'silent' -> ('e', 'i', 'l', 'n', 's', 't')
    Time Complexity: O(N * K log K), where K is the average word length.
    """
    groupes = {}
    
    for mot in liste_mots:
        # Tuple is mandatory because a list of characters is unhashable!
        signature = tuple(sorted(mot.lower()))
        if signature not in groupes:
            groupes[signature] = []
        groupes[signature].append(mot)
        
    return list(groupes.values())


def regrouper_anagrammes_comptage_optimal(liste_mots):
    """
    Approach 2 (State of the Art): 26-element character count tuple signature.
    Time Complexity: O(N * K) - No sorting overhead!
    """
    groupes = {}
    
    for mot in liste_mots:
        # Array of 26 zeros for a-z
        comptes = [0] * 26
        for char in mot.lower():
            if 'a' <= char <= 'z':
                comptes[ord(char) - ord('a')] += 1
                
        # Convert to tuple for hashing
        signature = tuple(comptes)
        if signature not in groupes:
            groupes[signature] = []
        groupes[signature].append(mot)
        
    return list(groupes.values())


# --- Tests & Demonstration ---
mots = ["listen", "silent", "enlist", "rat", "tar", "art", "evil", "vile", "live", "python"]

groupes_tri = regrouper_anagrammes_tri(mots)
groupes_opt = regrouper_anagrammes_comptage_optimal(mots)

print("=== Challenge 21: Grouping Anagrams ===")
print("Input words:", mots)
print("\nGrouped Anagrams (Sorted Signature):")
for g in groupes_tri:
    print(" ", g)

assert any(set(g) == {"listen", "silent", "enlist"} for g in groupes_tri)
assert any(set(g) == {"rat", "tar", "art"} for g in groupes_tri)
assert any(set(g) == {"evil", "vile", "live"} for g in groupes_tri)
assert any(set(g) == {"python"} for g in groupes_tri)
print("All anagram clusters validated successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Why the Tuple is the Linchpin of This Solution:**
   - Attempting `groupes[sorted(mot)] = ...` immediately fails with `TypeError: unhashable type: 'list'`.
   - `sorted(mot)` returns a **mutable list**. Converting it to `tuple(sorted(mot))` makes it immutable, hashable, and valid as a dictionary key.

2. **The 26-Element Frequency Tuple Technique ($O(N \times K)$):**
   - By creating a 26-tuple `(count_a, count_b, ..., count_z)`, we bypass sorting entirely.
   - For long text documents or large vocabularies, this reduces runtime from $O(N \times K \log K)$ down to linear $O(N \times K)$.


<H3><span style="font-weight: bold; color: #1a8bcdff;">Challenge 22: Mini Relational Join Engine (Hash Join: INNER & LEFT JOIN)</span>
<span style="background-color: #dc3545; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 8px;">Boss Challenge</span>
<span style="background-color: #6c757d; color: white; padding: 2px 8px; font-size: 0.65em; border-radius: 8px; margin-left: 5px;">Data Engineering • Hash Join • Pure Python SQL</span>
</H3>

1. <span style="font-weight: bold; color: #e56922ff;">Objective</span>:  
Implement a relational Hash Join algorithm in pure Python supporting both `INNER JOIN` and `LEFT JOIN` in optimal $O(N + M)$ time.

2. <span style="font-weight: bold; color: #e56922ff;">Context & Task</span>:  
A naive nested loop join `for r1 in t1: for r2 in t2:` costs $O(N \times M)$. Hash Joins index the right table into an in-memory hash table in $O(M)$ then probe it with the left table in $O(1)$ per row.

* **Input Example**:  
  Users: `[{'id': 1, 'name': 'Alice'}, {'id': 2, 'name': 'Bob'}, {'id': 3, 'name': 'Charlie'}]`  
  Orders: `[{'order_id': 101, 'user_id': 1, 'amount': 99.0}, {'order_id': 102, 'user_id': 1, 'amount': 15.0}, {'order_id': 103, 'user_id': 2, 'amount': 45.0}]`
* **Expected Output Example**:  
  INNER JOIN: Alice (2 orders), Bob (1 order). Charlie omitted.  
  LEFT JOIN: Alice (2 orders), Bob (1 order), Charlie (NULL order).


```python
def hash_inner_join(table_gauche, table_droite, cle_gauche, cle_droite):
    """
    INNER JOIN using Hash Join Algorithm in O(N + M):
    1. Build Phase: Index table_droite into a hash table (dict) by cle_droite.
    2. Probe Phase: Scan table_gauche and probe the hash table in O(1).
    """
    # Phase 1: Build index
    index_droite = {}
    for ligne in table_droite:
        cle = ligne.get(cle_droite)
        if cle is not None:
            if cle not in index_droite:
                index_droite[cle] = []
            index_droite[cle].append(ligne)
            
    # Phase 2: Probe index
    resultat = []
    for ligne_gauche in table_gauche:
        valeur_cle = ligne_gauche.get(cle_gauche)
        if valeur_cle in index_droite:
            for ligne_droite in index_droite[valeur_cle]:
                # Combine both records (union)
                resultat.append({**ligne_gauche, **ligne_droite})
                
    return resultat


def hash_left_join(table_gauche, table_droite, cle_gauche, cle_droite):
    """
    LEFT JOIN using Hash Join Algorithm in O(N + M):
    Guarantees every row from table_gauche appears at least once,
    padding right columns with None when there is no matching record.
    """
    # Build phase
    index_droite = {}
    toutes_cles_droite = set()
    for ligne in table_droite:
        toutes_cles_droite.update(ligne.keys())
        cle = ligne.get(cle_droite)
        if cle is not None:
            if cle not in index_droite:
                index_droite[cle] = []
            index_droite[cle].append(ligne)
            
    # Template of null values for right table keys
    null_droite = {k: None for k in toutes_cles_droite if k != cle_droite}
    
    # Probe phase
    resultat = []
    for ligne_gauche in table_gauche:
        valeur_cle = ligne_gauche.get(cle_gauche)
        if valeur_cle in index_droite:
            for ligne_droite in index_droite[valeur_cle]:
                resultat.append({**ligne_gauche, **ligne_droite})
        else:
            # Unmatched left row: pad with None
            resultat.append({**ligne_gauche, **null_droite})
            
    return resultat


# --- Tests & Demonstration ---
utilisateurs = [
    {"user_id": 1, "nom": "Alice",   "pays": "FR"},
    {"user_id": 2, "nom": "Bob",     "pays": "UK"},
    {"user_id": 3, "nom": "Charlie", "pays": "ES"}
]

commandes = [
    {"cmd_id": 101, "client_id": 1, "montant": 150.0},
    {"cmd_id": 102, "client_id": 1, "montant": 45.0},
    {"cmd_id": 103, "client_id": 2, "montant": 990.0},
    {"cmd_id": 104, "client_id": 9, "montant": 12.0} # Unknown customer
]

inner = hash_inner_join(utilisateurs, commandes, cle_gauche="user_id", cle_droite="client_id")
left = hash_left_join(utilisateurs, commandes, cle_gauche="user_id", cle_droite="client_id")

print("=== Challenge 22: Relational Hash Join ===")
print("INNER JOIN Results (Matched rows only):")
for r in inner:
    print(" ", r)

print("\nLEFT JOIN Results (All users preserved):")
for r in left:
    print(" ", r)

assert len(inner) == 3  # 2 orders for Alice, 1 for Bob
assert len(left) == 4   # 2 for Alice, 1 for Bob, 1 for Charlie with None values
print("All SQL relational joins validated successfully!")
```

#### Deep Analysis & Key Takeaways

1. **Hash Join ($O(N + M)$) vs. Naive Nested Loops ($O(N \times M)$):**
   - A nested loop join `for u in users: for c in orders: if u['id'] == c['user_id']` checks $N \times M$ pairs.
   - For 100,000 users and 1,000,000 orders, nested loops execute **100 billion checks**!
   - Hash Join indexes the right table in $O(M)$ memory then probes in $O(1)$, completing the entire join in $1,100,000$ operations!

2. **Why Big Data Engines (Spark, PostgreSQL, Snowflake) Use Hash Join:**
   - Hash Join is the primary join algorithm implemented by modern relational engines for equi-joins (`ON a.id = b.id`).
